## 3. Confirmatory Qwen result

For item $t$, condition $j$, and rollout $r$, $e_{tjr}=1$ denotes an error and $p_{tj}$ its latent error propensity. With $\mu_j=\mathbb{E}[p_{tj}]$, accuracy is $1-\mu_j$. Rescue and damage are the cross-policy masses $\mathbb{E}[p_{t0}(1-p_{tj})]$ and $\mathbb{E}[(1-p_{t0})p_{tj}]$, estimated with all four baseline--treatment rollout cross-products; rescue minus damage equals the accuracy change.

$G_j=\mathbb{E}[p_{t0}^2-p_{t0}p_{tj}]=O_{0j}-O_{00}$ compares the baseline--controller pair oracle with repeated baseline, so it combines mean competence change with covariance change. $C_j=G_j-\mu_0(\mu_0-\mu_j)=\operatorname{Var}(p_{t0})-\operatorname{Cov}(p_{t0},p_{tj})$ removes the mean-competence component and is the primary Q1 confirmatory endpoint.

$D_j=\mathbb{E}[(p_{t0}-p_{tj})^2]$ is the non-directional distance between error-propensity profiles, estimated unbiasedly from two independent rollout banks; a finite-sample estimate may be negative. **D measures movement, not damage.**

All holdout items remain in frozen manifest order. The upper panel shows the observed two-rollout error proportion rather than pretending each item has a deterministic state. Invalid or unevaluable terminal rows remain errors. The lower panels preserve the stochastic cross-rollout decomposition and show every prospective random control.
