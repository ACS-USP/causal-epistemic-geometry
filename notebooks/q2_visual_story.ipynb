## 0. What Q1 left unanswered

Q1 established a narrow causal fact: a prospectively fixed intervention can change *which* CRUXEval items Qwen3-8B fails, beyond repeated sampling and matched random directions. It did not show that different causal movements are organized relative to one another.

Q2 asks the relational question: **if two interventions are close before semantic correctness is observed, do they cause close blind-spot profiles afterward?** A single successful controller cannot answer this. We need a fixed intervention space, many controllers, outcome-free candidate geometries, and inference that respects shared controllers.


## 1. The intervention laboratory

Four source families measured at two standardized locations produce eight source directions in the 4,096-dimensional residual stream. Their rank-8 orthonormal span defines the laboratory. Forty directions were sampled once inside it; 31 prospectively passed the safety gate in both matched shells and became the immutable V4.1 bank.

The eight source directions **define the space**. The 31 tested directions **sample new points inside it**. They are not 31 named semantic concepts.

## 2. What counts as blind-spot distance?

For controller pair $i,j$, item $x$, and rollout block $r$, let $e_{i,x,r}=1$ denote semantic error and $d_{x,r}=e_{i,x,r}-e_{j,x,r}$. The frozen superpopulation estimator is

$$
\widehat D^{shape}_{ij}=\frac{N}{N-1}\left[\frac1N\sum_x d_{x,0}d_{x,1}-\left(\frac1N\sum_xd_{x,0}\right)\left(\frac1N\sum_xd_{x,1}\right)\right].
$$

The cross-rollout product estimates squared propensity difference; subtracting the product of mean differences removes global competence shift; $N/(N-1)$ converts the fixed-panel variance to the registered item-superpopulation target. The population target is non-negative, but its unbiased finite-sample estimate may be negative and is never clipped.

This is the critical distinction from Q1's total profile movement $D$: Q2's primary $D^{shape}$ asks whether the *pattern across items* differs after mean shift is removed.

## 3. Three pre-outcome geometries

The notebook does not assume one geometry is true. It compares three registered answers to “when are two interventions close?”

- **A0 — coordinate geometry:** $A0_{ij}=1-c_i^\top c_j$ for unit coefficients in the fixed span. It uses orientation only.
- **A1 — activation-statistical geometry:** a regularized covariance-whitened angular dissimilarity. It adds global anisotropy/correlation of label-free natural activations. It is not a learned manifold.
- **A2 — finite-response geometry:** a baseline-centered angle reconstructed from Jensen–Shannon distances among full-vocabulary responses on 12 label-free probes at four checkpoints. It is pre-semantic-outcome but uses finite downstream model response; it is not Fisher, a JVP, a pullback tensor, or a Riemannian metric.

A2 contains more downstream information, but more information does not guarantee a stronger semantic relational association.


## 4. Primary result: moderate relational association

Read each panel from left to right: as intervention-distance rank increases, the center of the blind-spot-rank distribution generally rises. The wide IQR and 10–90% intervals are equally important: the same intervention-distance decile contains broadly different blind-spot distances. Ranks are shown because Spearman rank association is the frozen statistic.

The effect size and the inferential evidence answer different questions:

- **Strength:** shell-specific $\rho$ values of 0.44–0.57 are moderate and leave substantial pair-specific variation unresolved.
- **Evidence against random label correspondence:** the frozen dependence-aware QAP gives maxT-adjusted $p=0.00002$ for all three geometries.

A tiny QAP p-value establishes evidence for a relational association; it does not make that association near-deterministic.

## 5. Why the classification is G2

A geometry qualifies only if both shell correlations are positive, the aggregate is at least 0.20, the frozen bootstrap lower bound is positive, every delete-one-controller result preserves sign, and the maxT-adjusted QAP p-value is at most 0.05. A0, A1, and A2 all qualify.

## 9. What Q2 can claim

**Strongest supported claim:** Within a prospectively fixed intervention subspace of Qwen3-8B, pre-outcome angular geometry was substantially associated with the subsequently observed pairwise rank structure of causal blind-spot distances across two matched amplitudes. Observed Spearman correlations were moderate ($\rho=0.44$–$0.57$), leaving substantial pair-specific variation unresolved.

QAP establishes evidence against random controller-label correspondence, while $\rho$ describes association strength. The positive item-bootstrap lower bounds, delete-one-controller sign stability, independent forensic agreement, and complete 37,800/37,800-row execution support confidence in the tested-bank association; A0, A1, and A2 are related candidate geometries, not three independent replications.

Separately, increasing intervention amplitude increased blind-spot displacement in all 31 tested directions. The finite-response A2 geometry also carried relational signal, but did not outperform simpler static intervention geometries.

The terminal mechanical classification is `Q2_V4_1_G2`; the independent radial classifications are `RS+` and `RT+`; the forensic audit is clean with maximum discrepancy 0.0.


## 10. What Q2 cannot claim

This notebook does not establish precise prediction for an individual controller pair; out-of-sample generalization to a fresh controller bank; model-, architecture-, layer-, benchmark-, or task-general geometry; a discovered activation manifold; a Fisher, pullback, Riemannian, or geodesic metric; global smoothness; linear or continuous dose response; A0 as the true geometry; universal inferiority of A2; transfer to LiveCodeBench; or deployable collective utility. The 31 controllers describe the prospectively sampled, safety-conditioned population tested here, not all directions in the span.


## 11. Boundary to future Q3

Q2 establishes a moderate pre-outcome relational association inside the frozen design. It does **not** establish held-out forecasting on new controllers, nor show that a non-oracle system can exploit profile diversity to improve decisions.
